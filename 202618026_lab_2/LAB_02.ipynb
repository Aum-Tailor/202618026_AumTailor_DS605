```python
mat_A = np.array([[4, 7], 
                  [2, 6]])
                  
mat_B = np.array([[1, 2], 
                  [3, 4]])

print("Addition:\n", mat_A + mat_B)
print("\nElement-wise Multiplication:\n", mat_A * mat_B)
print("\nMatrix Multiplication:\n", mat_A @ mat_B)

print("\nTranspose of mat_A:\n", mat_A.T)

det_A = np.linalg.det(mat_A)
print(f"\nDeterminant of mat_A: {det_A:.2f}")

if det_A != 0:
    inv_A = np.linalg.inv(mat_A)
    print("\nInverse of mat_A:\n", inv_A)
    is_identity = np.allclose(mat_A @ inv_A, np.eye(mat_A.shape[0]))
    print(f"\nIs mat_A @ mat_A^-1 equal to the Identity matrix? {is_identity}")
```

```python
import matplotlib.pyplot as plt
chosen_mean = 50.0
chosen_std = 15.0

rng = np.random.default_rng(42)
data = rng.normal(loc=chosen_mean, scale=chosen_std, size=1000)

sample_mean = np.mean(data)
sample_std = np.std(data)

print(f"Chosen Mean: {chosen_mean} | Sample Mean: {sample_mean:.2f}")
print(f"Chosen Std Dev: {chosen_std} | Sample Std Dev: {sample_std:.2f}")

plt.hist(data, bins=30, edgecolor='black', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=2, label=f'Sample Mean: {sample_mean:.2f}')
plt.title('Normal Distribution Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\AUM\Downloads\titanic\train.csv")

print("--- head() ---")
print(df.head())

print("\n--- tail() ---")
print(df.tail())

print("\n--- shape ---")
print(df.shape)

print("\n--- columns ---")
print(df.columns)

print("\n--- info() ---")
df.info() 

print("\n--- describe() ---")
print(df.describe())

print("\n--- loc Example ---")
print(df.loc[0:2, ['Name', 'Survived', 'Age']])

print("\n--- iloc Example ---")
print(df.iloc[0:3, 0:3])

print("\n--- Difference between loc and iloc ---")
print("loc is label-based: selects rows and columns by their actual index labels and column names.")
print("iloc is integer-based: selects rows and columns by their numerical positions (0-indexed).")
```

```python
# 1. Male passengers older than 50: how many?
q1 = df.query("Sex == 'male' and Age > 50")
print(f"Male passengers older than 50: {len(q1)}")

# 2. Female first-class passengers: how many, and what percentage survived?
q2 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
survived_pct = q2['Survived'].mean() * 100
print(f"Female first-class passengers: {len(q2)}")
print(f"Percentage survived: {survived_pct:.2f}%")

# 3. Age 20-40, Fare above overall median, and survived: how many?
overall_median_fare = df['Fare'].median()
q3 = df[(df['Age'] >= 20) & (df['Age'] <= 40) & (df['Fare'] > overall_median_fare) & (df['Survived'] == 1)]
print(f"Age 20-40, Fare above overall median, and survived: {len(q3)}")

# 4. Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
q4 = df.query("SibSp == 0 and Parch == 0 and Age < 30 and Survived == 0")
print(f"Travelling alone, age < 30, and did not survive: {len(q4)}")

# 5. Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?
s_median_fare = df[df['Embarked'] == 'S']['Fare'].median()
q5 = df[(df['Embarked'] == 'S') & (df['Pclass'].isin([2, 3])) & (df['Fare'] > s_median_fare)]
print(f"Embarked='S', Pclass 2/3, Fare > Southampton median: {len(q5)}")
```

```python
# 1. Survival rate by Sex
print("--- Survival rate by Sex ---")
print(df.groupby('Sex')['Survived'].mean())

# 2. Survival rate by Pclass
print("\n--- Survival rate by Pclass ---")
print(df.groupby('Pclass')['Survived'].mean())

# 3. Average Age and Fare by Pclass
print("\n--- Average Age and Fare by Pclass ---")
print(df.groupby('Pclass')[['Age', 'Fare']].mean())

# 4. Passenger count and survival rate by Sex-Pclass
print("\n--- Passenger count and survival rate by Sex-Pclass ---")
print(df.groupby(['Sex', 'Pclass'])['Survived'].agg(['count', 'mean']))

# 5. Passenger count, average Fare, and survival rate by Embarked
print("\n--- Passenger count, average Fare, and survival rate by Embarked ---")
print(df.groupby('Embarked').agg(
    Passenger_Count=('PassengerId', 'count'),
    Average_Fare=('Fare', 'mean'),
    Survival_Rate=('Survived', 'mean')
))
```
